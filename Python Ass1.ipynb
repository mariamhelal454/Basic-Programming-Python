```python
# 1. Password Validator
Password = input("Enter a Password: ")
digit = False
Upper = False

for i in Password:
    if i.isdigit():
        digit = True
    elif i.isupper():
        Upper = True

if len(Password) >= 8 and digit and Upper:
    print("Valid Password")
else:
    print("Invalid Password")          
```

```python
# 2. Count Characters in a String
sentence = input("Enter a sentence")

vowels_count = 0
consonants_count = 0
digits_count = 0
spaces_count = 0

Vowels = "aeiouAEIOU"

for i in sentence:
    if i in Vowels:
        vowels_count += 1
    elif i.isalpha():
        consonants_count += 1
    elif i.isdigit():
        digits_count += 1
    elif i == " ":
        spaces_count += 1


print(f"Vowels: {vowels_count}")
print(f"Consonants: {consonants_count}")
print(f"Digits: {digits_count}")
print(f"Spaces: {spaces_count}")
```

```python
# 3. Student Grades List
grades = [75, 48, 90, 62, 35, 88, 51]

passed = 0
failed = 0

for grade in grades:
    if grade >= 50 :
        passed += 1
    else:
        failed += 1

print(f"Passed student: {passed}")   
print(f"Failed student: {failed}")      
```

```python
# 4. Find the Largest and Smallest Number
numbers = [15, 42, 8, 91, 23, 4, 55]

largest_num = numbers[0]
smallest_num = numbers[0]

for num in numbers:
    if num > largest_num:
        largest_num = num
    elif num < smallest_num:
        smallest_num = num    


print(f"Largest number: {largest_num}")
print(f"Smallest number: {smallest_num}")
```

```python
# 5. Word Frequency Dictionary
sentence = input("Enter the sentence : ")
words = sentence.lower().split()

Word_Frequency ={}

for word in words:
    if word in Word_Frequency:
        Word_Frequency[word] += 1
    else:
        Word_Frequency[word] = 1

print(Word_Frequency)
```

```python
# 6. Remove Duplicate Values
numbers = [5, 2, 5, 8, 2, 10, 8, 12]
unique_numbers = []

for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)

print(unique_numbers)
```

```python
# 7. Group Words by First Letter
words = ["apple", "banana", "avocado", "blueberry", "cherry", "apricot"]
grouped_words = {}

for word in words:
    first_letter = word[0].lower()
    if first_letter in grouped_words:
        grouped_words[first_letter].append(word)
    else:
        grouped_words[first_letter] = [word]

print(grouped_words)
```

```python
# 8. Shopping Cart
cart = {
"Laptop": 25000,
"Mouse": 500,
"Keyboard": 1200,
"Headphones": 900
}

total_price = 0
items_above_1000 = 0

for product , price in cart.items():
    total_price += price
    if price > 1000:
        items_above_1000 += 1

print(f"Total Price: {total_price}")
print(f"Products costing more than 1000: {items_above_1000}")

if total_price > 25000:
    print("Expensive Order")
else:
    print("Normal Order")
```

```python
# 9. Common Values Between Two Lists
list1 = [1, 2, 3, 4, 5, 8]
list2 = [3, 4, 5, 6, 7, 8]

common_values =[]

for item in list1:
    if item in list2 and item not in common_values:
        common_values.append(item)

common_tuple = tuple(common_values)
print(common_tuple)

```

```python
# 10. Student Management Task
students = {
"Ahmed": [70, 80, 90],
"Sara": [95, 88, 92],
"Omar": [40, 55, 45],
"Mona": [60, 75, 70]
}  

passed_students = []
failed_students = [] 

for student_name , grades in students.items():
    total_grades = 0
    for grade in grades:
        total_grades += grade

    average = total_grades / len(grades)
    print(f"{student_name} - Average: {average}")

    if average >= 50:
        print("Passed")
        passed_students.append(student_name)
    else:
        print("Failed")
        failed_students.append(student_name)

print("Passed Students:", passed_students)
print("Failed Students:", failed_students)        

```
